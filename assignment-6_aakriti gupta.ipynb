Question 1

```python
import numpy as np

temperature = np.array([25, 28, 31, 35, 38, 27, 33, 40])

corrected_temperature = temperature + 2
print(corrected_temperature)

fahrenheit = (9/5) * corrected_temperature + 32
print(fahrenheit)

above_32 = corrected_temperature[corrected_temperature > 32]
print(above_32)

count = np.sum(corrected_temperature > 32)
print(count)
```

Question 2

```python
import numpy as np

steps = np.array([
    [5000, 6200, 7100],
    [8000, 7500, 9000],
    [4500, 5100, 4800],
    [9000, 8500, 9500]
])

print(np.sum(steps))

print(np.mean(steps))

print(np.max(steps))
print(np.min(steps))

print(np.sum(steps, axis=0))

print(np.sum(steps, axis=1))

max_position = np.unravel_index(np.argmax(steps), steps.shape)
print(max_position)
```

```python
import numpy as np

original = np.array([1, 2, 3, 4, 5, 6])

subset = original[1:5]
print(subset)

subset[0] = 999
print(original)
print(subset)

copied_array = original[1:5].copy()
copied_array[0] = 500
print(original)
print(copied_array)

matrix = np.arange(1, 13).reshape(3, 4)
print(matrix)

print(matrix[0])
print(matrix[-1])
print(matrix[:, 1])
print(matrix[1:3, 2:4])

flattened = matrix.flatten()
raveled = matrix.ravel()

print(flattened)
print(raveled)

raveled[0] = 100
print(matrix)

flattened[1] = 200
print(matrix)

print(matrix.shape)
print(matrix.ndim)
print(matrix.size)
print(matrix.dtype)
```

Question 4

```python
import numpy as np

X = np.array([
    [6, 70, 3],
    [5, 50, 6],
    [8, 80, 2],
    [4, 30, 8]
])

y = np.array([40, 65, 30, 85])

print(X.shape)
print(X.ndim)

print(X.T)

print(X.T @ X)

inverse = np.linalg.inv(X.T @ X)
print(inverse)

beta = np.linalg.inv(X.T @ X) @ X.T @ y
print(beta)

new_user = np.array([5, 40, 7])
predicted_score = new_user @ beta
print(predicted_score)
```
